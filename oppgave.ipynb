## Magnitudeskalaen

Magnituden $M$ til et jordskjelv er en logaritme:

- Når $M$ øker med 1, blir utslaget (amplituden) på en seismograf **10 ganger** større.
- Energien som frigjøres, blir omtrent $10^{1{,}5} \approx 32$ ganger større.

Forholdet mellom energien i to jordskjelv er derfor

$$\frac{E_1}{E_2} = 10^{1{,}5\,(M_1 - M_2)}$$

## Hente data

USGS sitt API svarer i formatet **GeoJSON**. Hvert jordskjelv ligger i listen `features`, med magnitude, sted og tid under `properties`, og koordinater (lengdegrad, breddegrad, dybde) under `geometry`.

```python
import requests

def hent_jordskjelv(fra, til, min_magnitude=4.5):
    svar = requests.get(
        "https://earthquake.usgs.gov/fdsnws/event/1/query",
        params={"format": "geojson", "starttime": fra, "endtime": til, "minmagnitude": min_magnitude},
        timeout=60,
    )
    return svar.json()


data = hent_jordskjelv("2025-01-01", "2025-02-01")
print("Antall jordskjelv:", len(data["features"]))
print(data["features"][0]["properties"]["mag"], data["features"][0]["properties"]["place"])
```
